```python
import pandas as pd
```

```python
walking = pd.read_csv("walking1.txt", sep="\t")
```

```python
walking["time_s"] = (walking["time"] - walking["time"].iloc[0]) / 1000000

walking.head()

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(walking["time_s"], walking["ax"], label="ax")
plt.plot(walking["time_s"], walking["ay"], label="ay")
plt.plot(walking["time_s"], walking["az"], label="az")

plt.xlabel("Time (seconds)")
plt.ylabel("Acceleration (g)")
plt.title("Walking1 - Accelerometer Data")
plt.legend()
plt.grid()

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(walking["time_s"], walking["gx"], label="gx")
plt.plot(walking["time_s"], walking["gy"], label="gy")
plt.plot(walking["time_s"], walking["gz"], label="gz")

plt.xlabel("Time (seconds)")
plt.ylabel("Angular velocity (degrees/second)")
plt.title("Walking1 - Gyroscope Data")
plt.legend()
plt.grid()

plt.show()
```

```python
# PLOT MAGNETOMETER

# Prefer the already-loaded walking dataframe; fall back to trials only if it exists
if "trials" in globals() and isinstance(trials, dict) and "walking1" in trials:
    walking = trials["walking1"]
else:
    walking = globals()["walking"]

plt.figure(figsize=(10, 5))

plt.plot(
    walking["time_s"],
    walking["mx"],
    label="mx"
)

plt.plot(
    walking["time_s"],
    walking["my"],
    label="my"
)

plt.plot(
    walking["time_s"],
    walking["mz"],
    label="mz"
)

plt.xlabel("Time (seconds)")
plt.ylabel("Magnetic field")
plt.title("Walking1 - Magnetometer Data")

plt.legend()
plt.grid()

plt.show()
```
